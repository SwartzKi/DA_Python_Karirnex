## Soal 2: Jumlah Transaksi per Kategori Mobil

Hitung ada berapa banyak transaksi untuk masing-masing kategori mobil (kolom `category`), diurutkan dari yang paling banyak.

**Bonus (opsional):** buat bar chart dari hasilnya.


```python
# Hitung jumlah transaksi per kategori mobil (category), urutkan dari yang paling banyak
transaksi_kategori = df['category'].value_counts()
print(transaksi_kategori, '\n')

# (opsional) buat bar chart dari hasilnya
# Untuk membuat chart dibutuhkan library matplotlib.pyplot
import matplotlib.pyplot as plt

# plot digunakan untuk memanggil library matplotlib.pyplot
# kind digunakan untuk menentukan jenis chart yang digunakan
# figsize digunakan untuk menentukan ukuran chart (lebar, tinggi)
# title digunakan untuk memberikan judul pada chart
# x dan y label digunakan untuk title pada sumbu x dan y
# xticks digunakan agar label pada sumbu x tidak muncul secara vertikal
# tight_layout digunakan untuk mengatur jarak antara chart dan label

transaksi_kategori.plot(kind="bar", figsize=(5, 3), title="Total Transaksi per Kategori")
plt.ylabel("Total Transaksi")
plt.xlabel("Kategori")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

```

## Soal 4: Bersihkan dan Siapkan Data

Sebelum data ini dipakai buat analisis lebih lanjut, kamu diminta membersihkan dan merapikannya dulu. Lakukan langkah-langkah berikut secara berurutan, dan simpan hasilnya ke `df` (menimpa data yang lama):

1. Cek apakah ada **missing value** di tiap kolom. Kalau kolom `trade_in` ada yang kosong, isi dengan `0`.
2. Cek apakah ada **baris duplikat** (baik duplikat penuh maupun duplikat `order_id`). Kalau ada, hapus baris duplikatnya.
3. Standarisasi kolom `payment_type`: hilangkan spasi berlebih di depan/belakang, dan samakan kapitalisasinya jadi Title Case (contoh: `"kredit"` jadi `"Kredit"`).
4. Ubah kolom `sales_date` yang masih berupa teks jadi tipe data tanggal (`datetime`).
5. Setelah semua langkah di atas selesai, tampilkan `df.info()` buat membuktikan datanya sudah lebih rapi (nggak ada missing value yang bermasalah, tipe data `sales_date` sudah benar).


```python
# # 1. Cek missing value, lalu isi trade_in yang kosong dengan 0
df["trade_in"] = df["trade_in"].fillna(0)

# Cek data trade_in
# dilakukan sebelum cleaning untuk memeriksa apakah ada data kosong
# dan dilakukan setelah cleaning untuk memastikan data sudah terisi
# df['trade_in'].head(5)



# # 2. Cek dan hapus baris duplikat (baik duplikat penuh maupun duplikat order_id)
df = df.drop_duplicates()

# Cek data trade_in
# dilakukan sebelum cleaning untuk memeriksa apakah ada data duplikat
# pengecekan dilakukan dengan memeriksa jumlah data
# dan dilakukan setelah cleaning untuk memastikan data sudah bersih
# dilakukan dengan memeriksa apakah jumlah data berkurang
# jika tidak berkurang berarti data sudah bersih dari awal
# print("Jumlah baris setelah hapus duplikat:", len(df))



# # 3. Standarisasi kolom payment_type (strip + title case)
# Cek isi sebelum standarisasi
# df['payment_type'].unique()

df['payment_type'] = df['payment_type'].str.strip().str.title()
# Cek isi setelah standarisasi
df['payment_type'].unique()



# # 4. Ubah kolom sales_date jadi tipe data datetime
# Cek sebelum ubah tipe data
# print(df["sales_date"].dtype)
# df["sales_date"].head()

# Mengubah tipe data tanggal menjadi datetime menggunakan pandas
# error='coerce' digunakan untuk mengubah nilai yang tidak dapat diubah
# menjadi datetime menjadi NaT (Not a Time)
# penggunaan coerce opsional. hanya sebagai antisipasi jika terdapat data yang tidak valid
# tanpa coerce akan terjadi error. dengan coerce perintah akan terus berjalan tanpa error

df["sales_date"] = pd.to_datetime(df["sales_date"], errors="coerce")

print(df["sales_date"].dtype)
df["sales_date"].head()


# # 5. Tampilkan df.info() untuk membuktikan data sudah lebih rapi
df.info()

```

## Soal 5: Kombinasi Cabang & Kategori dengan Pendapatan Tertinggi

Manajemen mau tahu: **kombinasi cabang dan kategori mobil apa** yang menghasilkan pendapatan (`total_sales`) paling besar? Ini penting buat mereka menentukan cabang mana yang perlu ditambah stok kategori tertentu.

Kerjakan langkah-langkah berikut:

1. Ambil hanya transaksi dengan status `"completed"` (transaksi yang batal/refund tidak dihitung sebagai pendapatan).
2. Kelompokkan data berdasarkan **kombinasi** `branch` dan `category` sekaligus (bukan salah satu aja), lalu jumlahkan `total_sales`-nya.
   - *Petunjuk: `groupby()` bisa menerima lebih dari satu kolom sekaligus, dengan cara `df.groupby(["kolom1", "kolom2"])`.*
3. Urutkan hasilnya dari pendapatan terbesar, lalu tampilkan **5 kombinasi cabang + kategori** dengan pendapatan tertinggi.
4. Buat bar chart horizontal dari 5 kombinasi teratas tersebut, biar gampang dipresentasikan.

**Tantangan tambahan (opsional, buat yang mau nyoba lebih jauh):** dari 5 kombinasi teratas itu, hitung juga berapa **rata-rata harga per unit** (`price`) untuk masing-masing kombinasi tersebut.


```python
# # (opsional) Tantangan tambahan: rata-rata harga per unit untuk 5 kombinasi teratas

# # karena yang dibutuhkan kombinasi jadi yang saya tulis hanya satu tidak seperti soal diatas

# Mengambil total sales dan total quantity
avg_kombinasi = (
    df.groupby(["branch", "category"])
    .agg(
        total_sales = ('total_sales', 'sum'),
        total_quantity = ('quantity', 'sum')
    )
)

# Menghitung rata-rata
avg_kombinasi["avg_sales"] = (
    avg_kombinasi["total_sales"] / avg_kombinasi["total_quantity"]
)

# Mengurutkan data
avg_kombinasi = avg_kombinasi.sort_values("avg_sales", ascending=False)

# Menampilkan data
avg_kombinasi.head(5)
```

```python
# # 1. Filter transaksi yang statusnya "completed"

transaksi_selesai = df[df["status"] == "completed"]
print("Jumlah transaksi completed:", len(transaksi_selesai))
transaksi_selesai.head(5)





# # 2. Groupby kombinasi branch dan category, jumlahkan total_sales

# Reset index agar branch bisa di grouping secara terpisah
# () digunakan setelah = agar kode bisa ditulis dengan baris jadi lebuh mudah dibaca
penjualan_per_kombinasi = (
    df.groupby(["branch", "category"])["total_sales"]
    .sum()
    .reset_index()
)

# Mengambil index branch agar dapat di grouping/branch
urutan_branch = (
    penjualan_per_kombinasi.groupby("branch")["total_sales"]
    .sum()
    .sort_values(ascending=False)
    .index
)

# Menyimpan branch sebagai urutan
# Categorical digunakan agar data yang telah disusun dapat disimpan dan digunakan kembali
penjualan_per_kombinasi["branch"] = pd.Categorical(
    penjualan_per_kombinasi["branch"],
    # Mengambil data yang telah disusun
    categories = urutan_branch,
    # Menyimpan data agar dapat digunakan kembali
    ordered = True
)

# Mengurutkan data berdasarkan total_sales
penjualan_per_kombinasi = (
    penjualan_per_kombinasi
    .sort_values(["branch", "total_sales"], ascending = [True, False])
)

# Menunjukkan hasil branch, category, dan total_sales
# print(penjualan_per_kombinasi)





# # 3. Urutkan dari terbesar, ambil 5 kombinasi teratas

# # Terbesar apa? per category? per total_sales? per branch?
# # karena bingung dan bisa jadi saya tampilkan ketiganya dengan cara yang serupa

# Dilakukan grouping terlebih dahulu
# Jika kolom yang digunakan lebih dari satu, tambah [] di dalam groupby
# Total sales dijumlahkan dan diurutkan dari yang terbesar

# agar tidak mengganggu kode pada nomor 2, saya membuat variabel baru

# total_sales terbesar
total_sales_tertinggi = (
    df.groupby(["branch", "category"])["total_sales"]
    .sum().sort_values(ascending=False)
)
total_sales_tertinggi.head(5)

# branch terbesar
branch_tertinggi = (
    df.groupby("branch")["total_sales"]
    .sum().sort_values(ascending=False)
)
branch_tertinggi.head(5)

# category terbesar
category_tertinggi = (
    df.groupby("category")["total_sales"]
    .sum().sort_values(ascending=False)
)
category_tertinggi.head(5)





# # 4. Buat bar chart horizontal dari 5 kombinasi teratas

# caranya sama seperti bar normal
# yang membedakan horizontal menggunakan barh
# total sales dipersingkat secara default dikarenakan nilai terlalu besar
# keterangan 1e11 = 1x10^11

# Chart total_sales_tertinggi
total_sales_tertinggi.head(5).sort_values().plot(
    kind="barh",
    figsize=(5, 3),
    title="Jumlah Total Sales Terbesar"
)
plt.xlabel("Total Sales")
plt.tight_layout()
plt.show()

# Chart branch_tertinggi
branch_tertinggi.head(5).sort_values().plot(
    kind="barh",
    figsize=(5, 3),
    title="Jumlah Total Sales Terbesar per Branch"
)
plt.xlabel("Total Sales")
plt.tight_layout()
plt.show()

# Chart category_tertinggi
category_tertinggi.head(5).sort_values().plot(
    kind="barh",
    figsize=(5, 3),
    title="Jumlah Total Sales terbesar per Kategori"
)
plt.xlabel("Total Sales")
plt.tight_layout()
plt.show()

# # Jika dilihat kembali yang dibutuhkan memang hanya branch & category
# # jadi seharusnya masing-masing branch dan category tidak perlu





# # (opsional) Tantangan tambahan: rata-rata harga per unit untuk 5 kombinasi teratas

# # karena yang dibutuhkan kombinasi jadi yang saya tulis hanya satu tidak seperti soal diatas

# Mengambil total sales dan total quantity
avg_kombinasi = (
    df.groupby(["branch", "category"])
    .agg(
        total_sales = ('total_sales', 'sum'),
        total_quantity = ('quantity', 'sum')
    )
)

# Menghitung rata-rata
avg_kombinasi["avg_sales"] = (
    avg_kombinasi["total_sales"] / avg_kombinasi["total_quantity"]
)

# Mengurutkan data
avg_kombinasi = avg_kombinasi.sort_values("avg_sales", ascending=False)

# Menampilkan data
avg_kombinasi.head(5)
```
